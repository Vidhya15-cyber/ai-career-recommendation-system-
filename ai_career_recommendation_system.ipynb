```python
df_clean = df.copy()
```

```python
print("Duplicates before cleaning:", df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

print("Duplicates after cleaning:", df_clean.duplicated().sum())
```

```python
print("Missing values before cleaning:")
print(df_clean.isnull().sum())
```

```python
numeric_columns = df_clean.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df_clean[column] = df_clean[column].fillna(df_clean[column].median())
```

```python
categorical_columns = df_clean.select_dtypes(include="object").columns

for column in categorical_columns:
    df_clean[column] = df_clean[column].fillna(df_clean[column].mode()[0])
```

```python
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
for column in categorical_columns:
    df_clean[column] = df_clean[column].astype(str).str.strip()
```

```python
print("Final Dataset Shape:", df_clean.shape)
df_clean.head()
```

```python
df_clean.to_csv("career_dataset_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
X_reg = df.drop(columns=["Career", "GPA"])

y_reg = df["GPA"]
```

```python
X_reg = pd.get_dummies(X_reg, drop_first=True)

X_reg = X_reg.astype(float)
```

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

print("Training Data:", X_train_reg.shape)
print("Testing Data:", X_test_reg.shape)
```

```python
reg_model = LinearRegression()
```

```python
reg_model.fit(X_train_reg, y_train_reg)
```

```python
y_pred_reg = reg_model.predict(X_test_reg)
```

```python
mse = mean_squared_error(y_test_reg, y_pred_reg)
r2 = r2_score(y_test_reg, y_pred_reg)

print("Mean Squared Error:", mse)
print("R2 Score:", r2)
```

```python
comparison = pd.DataFrame({
    "Actual GPA": y_test_reg.values,
    "Predicted GPA": y_pred_reg
})

print(comparison.head(10))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
X = df.drop(columns=["Career"])
y = df["Career"]
```

```python
X = pd.get_dummies(X, drop_first=True)

X = X.astype(float)
```

```python
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

career_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

```python
career_model.fit(X_train, y_train)
```

```python
y_pred = career_model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Career Prediction Accuracy:", accuracy)
```

```python
print("Classification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```
