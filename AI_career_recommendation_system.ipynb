```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df[numerical_columns].hist(figsize=(12, 8))
plt.tight_layout()
plt.show()
```

```python
print(df.columns)
```

```python
plt.figure(figsize=(10, 6))

df["Career"].value_counts().plot(kind="bar")

plt.title("Career Distribution")
plt.xlabel("Career")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

df["Field"].value_counts().plot(kind="bar")

plt.title("Field Distribution")
plt.xlabel("Field")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=df, x="Career", y="GPA")

plt.title("GPA Distribution Across Careers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    df[numerical_columns].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
sns.pairplot(df[numerical_columns])
plt.show()
```
